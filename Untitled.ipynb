### Dataset Justification

This notebook uses the **Victorian Era Authorship Attribution Dataset** (Gungor, 2018), sourced from the UCI Machine Learning Repository / Kaggle. The dataset contains **93,600 text fragments** drawn from novels by **50 different Victorian-era authors** (e.g. Charles Dickens, George Eliot, Arthur Conan Doyle).

**Why this dataset fits the brief:**
- **Scale**: 93,600 records comfortably exceeds the minimum 10,000-record requirement.
- **Relevance**: A book press building an authorship-style bot needs a model trained on genuine literary prose with a large, diverse author pool — this dataset provides exactly that.
- **Task fit**: Authorship attribution from text sequences is a canonical sequence-modelling problem, well suited to an RNN/LSTM architecture that can learn stylistic and syntactic patterns across a sequence of words/characters.
- **Real-world grounding**: Victorian literature is public-domain and widely studied, allowing the model's predictions to be sanity-checked against known authorial styles.

**Reference:**
Gungor, A. (2018). *Victorian Era Authorship Attribution*. Kaggle. Available at: https://www.kaggle.com/datasets/josephgpinto/victorian-era-authorship-attribution

```python
# Core libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Spark (required for analysis per assignment brief)
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, length, split, size

# Deep learning
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Dropout, Bidirectional
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.utils import to_categorical

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import classification_report, confusion_matrix

import warnings
warnings.filterwarnings('ignore')

# Initialise Spark session
spark = SparkSession.builder \
    .appName("VictorianAuthorshipAttribution") \
    .config("spark.driver.memory", "4g") \
    .getOrCreate()

print("Spark version:", spark.version)
print("TensorFlow version:", tf.__version__)
```

```python
# Load dataset into Spark DataFrame
df_spark = spark.read.csv(
    "Gungor 2018 Victorian Author Attribution data train.csv",
    header=True,
    inferSchema=True
)

# Basic inspection
print("Number of records:", df_spark.count())
print("Number of columns:", len(df_spark.columns))
df_spark.printSchema()
df_spark.show(5, truncate=80)
```

```python
# Check how many unique authors and record distribution per author
df_spark.groupBy("author").count().orderBy(col("count").desc()).show(50)

num_authors = df_spark.select("author").distinct().count()
print("Number of unique authors:", num_authors)
```

```python
# Add a word-count column and inspect text length distribution
df_spark = df_spark.withColumn("word_count", size(split(col("text"), " ")))

df_spark.select("word_count").describe().show()

# Word count distribution per author (top 10 authors by record count)
df_spark.groupBy("author") \
    .avg("word_count") \
    .orderBy(col("author")) \
    .show(50)
```

```python
# Convert a manageable sample to pandas for plotting (Spark handles the heavy analysis;
# pandas/matplotlib used only for visualisation as permitted)
df_sample = df_spark.sample(fraction=0.1, seed=42).toPandas()

plt.figure(figsize=(12,5))
sns.countplot(x="author", data=df_sample, order=df_sample["author"].value_counts().index)
plt.title("Sampled Record Count per Author")
plt.xlabel("Author ID")
plt.ylabel("Count")
plt.xticks(rotation=90)
plt.show()

plt.figure(figsize=(10,5))
sns.histplot(df_sample["word_count"], bins=50, kde=True)
plt.title("Distribution of Word Count per Text Fragment")
plt.xlabel("Word Count")
plt.show()
```

### EDA Summary

- Dataset contains 53,678 text fragments across 50 authors.
- Author distribution is imbalanced (author 8: 6,914 records; author 1: 912 records), which will be considered during model training (class weighting or stratified sampling).
- Text fragments show minimal length variance, as they were pre-segmented into near-uniform windows during dataset construction.
- No missing values were identified in the `text` or `author` columns (confirmed via Spark schema/count checks).

```python
# Convert Spark DataFrame to pandas for Keras tokenization (full dataset - manageable size)
df_pd = df_spark.select("text", "author").toPandas()

print(df_pd.shape)
df_pd.head()

# Encode author labels
label_encoder = LabelEncoder()
df_pd["author_encoded"] = label_encoder.fit_transform(df_pd["author"])
num_classes = df_pd["author_encoded"].nunique()
print("Number of classes:", num_classes)

# Tokenize text
MAX_VOCAB = 20000
MAX_LEN = 200

tokenizer = Tokenizer(num_words=MAX_VOCAB, oov_token="<OOV>")
tokenizer.fit_on_texts(df_pd["text"])

sequences = tokenizer.texts_to_sequences(df_pd["text"])
padded_sequences = pad_sequences(sequences, maxlen=MAX_LEN, padding="post", truncating="post")

print("Padded sequence shape:", padded_sequences.shape)
print("Vocabulary size used:", len(tokenizer.word_index))
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    padded_sequences,
    df_pd["author_encoded"].values,
    test_size=0.2,
    random_state=42,
    stratify=df_pd["author_encoded"]
)

y_train_cat = to_categorical(y_train, num_classes=num_classes)
y_test_cat = to_categorical(y_test, num_classes=num_classes)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train_cat.shape)
print("y_test shape:", y_test_cat.shape)
```

```python
VOCAB_SIZE = min(MAX_VOCAB, len(tokenizer.word_index) + 1)
EMBEDDING_DIM = 128

model = Sequential([
    Embedding(input_dim=VOCAB_SIZE, output_dim=EMBEDDING_DIM, input_length=MAX_LEN),
    Bidirectional(LSTM(128, return_sequences=True)),
    Dropout(0.3),
    Bidirectional(LSTM(64)),
    Dropout(0.3),
    Dense(64, activation="relu"),
    Dense(num_classes, activation="softmax")
])

model.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()
```

```python
early_stop = EarlyStopping(
    monitor="val_loss",
    patience=3,
    restore_best_weights=True
)

history = model.fit(
    X_train, y_train_cat,
    validation_split=0.1,
    epochs=15,
    batch_size=64,
    callbacks=[early_stop],
    verbose=1
)
```

```python
y_pred_probs = model.predict(X_test)
y_pred = np.argmax(y_pred_probs, axis=1)

print(classification_report(y_test, y_pred, target_names=[str(c) for c in label_encoder.classes_]))
```

```python
plt.figure(figsize=(16,14))
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, cmap="Blues", cbar=True,
            xticklabels=label_encoder.classes_,
            yticklabels=label_encoder.classes_)
plt.title("Confusion Matrix - Initial Model")
plt.xlabel("Predicted Author")
plt.ylabel("True Author")
plt.show()
```

### Evaluation Findings (Initial Model)

- Overall test accuracy: 42%.
- Strong performance on high-frequency authors (e.g. author 8: precision 0.84, recall 0.72, support 1,383).
- Very poor performance on low-frequency authors (e.g. authors 2, 11, 13, 22, 23: precision/recall/f1 of 0.00).
- Macro-average F1 (0.21) is far lower than weighted-average F1 (0.40), confirming the model is biased toward majority classes due to class imbalance.
- **Action taken:** the model will be retrained using class weighting so that under-represented authors contribute proportionally more to the loss function, aiming to improve macro-level performance without materially harming majority-class accuracy.

```python
from sklearn.utils.class_weight import compute_class_weight

class_weights_array = compute_class_weight(
    class_weight="balanced",
    classes=np.unique(y_train),
    y=y_train
)
class_weights = dict(enumerate(class_weights_array))

# Rebuild the model fresh (avoid continuing from biased weights)
model_v2 = Sequential([
    Embedding(input_dim=VOCAB_SIZE, output_dim=EMBEDDING_DIM, input_length=MAX_LEN),
    Bidirectional(LSTM(128, return_sequences=True)),
    Dropout(0.3),
    Bidirectional(LSTM(64)),
    Dropout(0.3),
    Dense(64, activation="relu"),
    Dense(num_classes, activation="softmax")
])

model_v2.compile(optimizer="adam", loss="categorical_crossentropy", metrics=["accuracy"])

early_stop_v2 = EarlyStopping(monitor="val_loss", patience=3, restore_best_weights=True)

history_v2 = model_v2.fit(
    X_train, y_train_cat,
    validation_split=0.1,
    epochs=15,
    batch_size=64,
    class_weight=class_weights,
    callbacks=[early_stop_v2],
    verbose=1
)
```

```python
test_loss_v2, test_accuracy_v2 = model_v2.evaluate(X_test, y_test_cat, verbose=1)
print(f"Model v2 Test Loss: {test_loss_v2:.4f}")
print(f"Model v2 Test Accuracy: {test_accuracy_v2:.4f}")

y_pred_probs_v2 = model_v2.predict(X_test)
y_pred_v2 = np.argmax(y_pred_probs_v2, axis=1)

print(classification_report(y_test, y_pred_v2, target_names=[str(c) for c in label_encoder.classes_]))
```

```python
model.save("model_v1_baseline.keras")
model_v2.save("model_v2_classweighted.keras")
print("Both models saved.")
```

```python
plt.figure(figsize=(16,14))
cm_v2 = confusion_matrix(y_test, y_pred_v2)
sns.heatmap(cm_v2, cmap="Oranges", cbar=True,
            xticklabels=label_encoder.classes_,
            yticklabels=label_encoder.classes_)
plt.title("Confusion Matrix - Class-Weighted Model (v2)")
plt.xlabel("Predicted Author")
plt.ylabel("True Author")
plt.show()
```

### Comparative Analysis: Baseline vs Class-Weighted Model

| Metric              | Baseline (v1) | Class-Weighted (v2) |
|----------------------|:--------------:|:--------------------:|
| Test Accuracy        | 0.42           | 0.19                 |
| Macro Avg F1          | 0.21           | 0.16                 |
| Weighted Avg F1       | 0.40           | 0.18                 |

**Discussion:**

Retraining with `class_weight="balanced"` was applied specifically to address the class imbalance identified during evaluation of the baseline model (author 8: 1,383 samples vs authors such as 2, 11, 13: under 100 samples each). Contrary to the intended outcome, class weighting **reduced performance across every metric** rather than improving it.

The most likely explanation is that full inverse-frequency weighting over-corrected: it forced the loss function to prioritise very rare authors so heavily that the model struggled to converge on a stable decision boundary for the majority of classes within the same training budget (15 epochs, `patience=3`). The baseline model's early stopping (epoch 8) captured a more favourable point in training before overfitting set in, whereas the class-weighted model may have needed a longer training schedule or a gentler weighting scheme to reach a comparable optimum.

**Conclusion:**

The baseline model (v1) is retained as the better-performing model for this dataset and task, achieving 42% test accuracy against a 45-class problem with severe imbalance. The retraining experiment demonstrates that naive class weighting is not a guaranteed fix for imbalance in sequence classification tasks, and highlights that model improvement strategies must be evaluated empirically rather than assumed to work.

**Recommendations for future work:**
- Apply a milder weighting scheme (e.g. square-root of inverse frequency) rather than full inverse-frequency balancing.
- Consider oversampling minority-class fragments instead of, or in addition to, loss reweighting.
- Explore focal loss, which down-weights easy (majority-class) examples rather than up-weighting rare ones.
- Increase training epochs/patience specifically for the reweighted model to allow convergence under the harder loss landscape.
